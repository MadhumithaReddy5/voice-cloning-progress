# 🎤 Input-Based Multi-Language Voice Cloning
## Your Voice + Your Text + 10 Indian Languages + Colloquial Style

### Features:
- Enter YOUR text (any language)
- Get YOUR voice speaking it in 10 Indian languages
- Natural, colloquial speech patterns
- Same voice consistency across all languages

```python
# Upload your voice sample
from google.colab import files
print("📁 Upload your voice sample:")
uploaded = files.upload()
voice_file = list(uploaded.keys())[0]
print(f"✅ Voice file: {voice_file}")
```

```python
# Get your input text
user_text = input("📝 Enter your text to be spoken in all languages: ")
print(f"📄 Processing: {user_text}")

# Make text colloquial
def make_colloquial(text):
    return text.replace('.', ', ').replace('।', ', ').replace('?', '? ').replace('!', '! ')

colloquial_text = make_colloquial(user_text)
print(f"🗣️ Colloquial version: {colloquial_text}")
```

```python
# Setup TTS model
import torch
from TTS.api import TTS

device = "cuda" if torch.cuda.is_available() else "cpu"
print(f"🚀 Using: {device}")

print("📥 Loading model...")
tts = TTS("tts_models/multilingual/multi-dataset/xtts_v2").to(device)
print("✅ Model ready!")
```

```python
# Language mapping
languages = {
    'Telugu': 'te',
    'Hindi': 'hi',
    'Tamil': 'ta', 
    'Kannada': 'kn',
    'Malayalam': 'ml',
    'Bengali': 'bn',
    'Gujarati': 'gu',
    'Marathi': 'mr',
    'Punjabi': 'pa',
    'Urdu': 'ur'
}

print(f"🌍 Will process {len(languages)} languages with your input text")
```

```python
# Generate voice clones for all languages
for language, lang_code in languages.items():
    print(f"\n🎯 Processing {language}...")
    
    output_file = f"{language}_your_voice.wav"
    
    try:
        tts.tts_to_file(
            text=colloquial_text,
            speaker_wav=voice_file,
            language=lang_code,
            file_path=output_file
        )
        print(f"✅ {language} complete: {output_file}")
    except Exception as e:
        print(f"❌ {language} failed: {e}")

print("\n🎉 ALL LANGUAGES PROCESSED!")
```

```python
# Download all files
import glob

print("📥 Downloading all voice files...")
for file in glob.glob("*_your_voice.wav"):
    files.download(file)
    print(f"Downloaded: {file}")

print("\n🏆 COMPLETE!")
print(f"Your text '{user_text}' is now available in your voice across 10 Indian languages!")
```
