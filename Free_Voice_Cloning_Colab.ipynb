## Step 2: Upload Your Voice Sample

```python
from google.colab import files
import os

print("📁 Upload your English voice sample (WAV/MP3):")
uploaded = files.upload()

# Get the uploaded file name
voice_file = list(uploaded.keys())[0]
print(f"✅ Voice sample uploaded: {voice_file}")
print(f"📊 File size: {len(uploaded[voice_file])} bytes")
```

## Step 3: Setup Voice Cloning System

```python
import torch
from transformers import pipeline
from googletrans import Translator
from TTS.api import TTS
import librosa
import soundfile as sf

# Check GPU availability
device = "cuda" if torch.cuda.is_available() else "cpu"
print(f"🚀 Using device: {device}")

# Language mapping
LANGUAGES = {
    "Telugu": "te",
    "Hindi": "hi",
    "Tamil": "ta", 
    "Kannada": "kn",
    "Malayalam": "ml",
    "Bengali": "bn"
}

print(f"🌍 Will process {len(LANGUAGES)} Indian languages")
print("📋 Languages:", list(LANGUAGES.keys()))
```

## Step 4: Load Models

```python
# Load Whisper for transcription
print("📥 Loading Whisper model for transcription...")
whisper_pipe = pipeline(
    "automatic-speech-recognition",
    model="openai/whisper-medium",
    chunk_length_s=30,
    device=0 if device == "cuda" else -1
)
print("✅ Whisper loaded!")

# Load XTTS for voice cloning
print("📥 Loading XTTS model for voice cloning...")
try:
    tts = TTS("tts_models/multilingual/multi-dataset/xtts_v2").to(device)
    print("✅ XTTS loaded successfully!")
    voice_cloning_available = True
except Exception as e:
    print(f"❌ XTTS failed to load: {e}")
    print("📋 Will use fallback TTS methods")
    voice_cloning_available = False

# Initialize translator
translator = Translator()
print("✅ Translator ready!")
```

## Step 5: Process Your Voice

```python
# Transcribe the uploaded audio
print("🔤 Transcribing your audio...")
result = whisper_pipe(voice_file)
english_text = result["text"]
print(f"📄 Transcribed text: {english_text}")

# Create output directory
os.makedirs("voice_clones", exist_ok=True)
print("📁 Output directory created")
```

## Step 6: Clone Your Voice in All Indian Languages

```python
successful_clones = []
failed_clones = []

for language_name, lang_code in LANGUAGES.items():
    print(f"\n🎯 Processing {language_name}...")
    
    try:
        # Translate text
        print(f"🔄 Translating to {language_name}...")
        translated = translator.translate(english_text, dest=lang_code)
        translated_text = translated.text
        print(f"✅ Translation complete")
        
        # Generate voice clone
        output_file = f"voice_clones/{language_name}_your_voice.wav"
        
        if voice_cloning_available:
            print(f"🎤 Cloning your voice for {language_name}...")
            tts.tts_to_file(
                text=translated_text,
                speaker_wav=voice_file,
                language=lang_code,
                file_path=output_file
            )
            print(f"✅ Voice cloned: {output_file}")
            successful_clones.append(language_name)
        else:
            # Fallback to gTTS
            print(f"📢 Using fallback TTS for {language_name}...")
            from gtts import gTTS
            fallback_tts = gTTS(text=translated_text, lang=lang_code, slow=False)
            fallback_file = f"voice_clones/{language_name}_fallback.mp3"
            fallback_tts.save(fallback_file)
            print(f"✅ Fallback audio: {fallback_file}")
            successful_clones.append(f"{language_name} (fallback)")
            
    except Exception as e:
        print(f"❌ Failed to process {language_name}: {e}")
        failed_clones.append(language_name)

print(f"\n🎉 Processing Complete!")
print(f"✅ Successful: {len(successful_clones)} languages")
print(f"❌ Failed: {len(failed_clones)} languages")

if successful_clones:
    print(f"\n📋 Successfully processed: {', '.join(successful_clones)}")
if failed_clones:
    print(f"\n⚠️ Failed to process: {', '.join(failed_clones)}")
```

## Step 7: Download Your Voice Clones

```python
import glob
from google.colab import files

print("📥 Downloading all generated voice files...")

# Download all generated files
voice_files = glob.glob("voice_clones/*")

if voice_files:
    for file_path in voice_files:
        try:
            files.download(file_path)
            print(f"✅ Downloaded: {file_path}")
        except Exception as e:
            print(f"❌ Failed to download {file_path}: {e}")
    
    print(f"\n🎉 SUCCESS! Downloaded {len(voice_files)} voice files")
    print("\n📋 What you got:")
    for file_path in voice_files:
        filename = file_path.split('/')[-1]
        print(f"   • {filename}")
else:
    print("❌ No voice files were generated")

print("\n🏆 Voice cloning complete!")
```
