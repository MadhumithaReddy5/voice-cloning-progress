```python
# Multi-language voice cloning setup
import torch
from TTS.api import TTS
from gtts import gTTS
import os

device = "cuda" if torch.cuda.is_available() else "cpu"
print(f"🚀 Using: {device}")

# Load XTTS model
print("📥 Loading multilingual model...")
tts = TTS("tts_models/multilingual/multi-dataset/xtts_v2").to(device)

# Get uploaded voice file
voice_file = list(uploaded.keys())[0]
print(f"✅ Voice sample: {voice_file}")
```

```python
# Sample texts in different Indian languages
texts = {
    'Telugu': ('te', 'అదే మరి, నీవు లినియర్ రిగ్రెషన్ చూసినప్పుడు, అక్కడ మనం ఒక నంబర్ ప్రిడిక్ట్ చేయడానికి ప్రయత్నిస్తాము।'),
    'Hindi': ('hi', 'नमस्कार, मैं आपका वॉयस क्लोन हूं। मशीन लर्निंग बहुत दिलचस्प विषय है।'),
    'Tamil': ('ta', 'வணக்கம், நான் உங்கள் குரல் நகல். இயந்திர கற்றல் மிகவும் சுவாரஸ்யமான விஷயம்.'),
    'Kannada': ('kn', 'ನಮಸ್ಕಾರ, ನಾನು ನಿಮ್ಮ ಧ್ವನಿ ಕ್ಲೋನ್. ಯಂತ್ರ ಕಲಿಕೆ ಬಹಳ ಆಸಕ್ತಿದಾಯಕ ವಿಷಯ.'),
    'Malayalam': ('ml', 'നമസ്കാരം, ഞാൻ നിങ്ങളുടെ വോയ്സ് ക്ലോൺ ആണ്. മെഷീൻ ലേണിംഗ് വളരെ രസകരമായ വിഷയമാണ്.'),
    'Bengali': ('bn', 'নমস্কার, আমি আপনার ভয়েস ক্লোন। মেশিন লার্নিং খুবই আকর্ষণীয় বিষয়।'),
    'Gujarati': ('gu', 'નમસ્કાર, હું તમારો વૉઇસ ક્લોન છું. મશીન લર્નિંગ ખૂબ જ રસપ્રદ વિષય છે.'),
    'Marathi': ('mr', 'नमस्कार, मी तुमचा व्हॉइस क्लोन आहे. मशीन लर्निंग हा अतिशय मनोरंजक विषय आहे.')
}

print(f"🌍 Processing {len(texts)} Indian languages...")
```

```python
# Generate voice clones for all languages
for language, (lang_code, text) in texts.items():
    print(f"\n🎯 Processing {language}...")
    
    # Voice cloning (your voice)
    clone_file = f"{language}_your_voice.wav"
    try:
        tts.tts_to_file(
            text=text,
            speaker_wav=voice_file,
            language=lang_code,
            file_path=clone_file
        )
        print(f"✅ Your voice in {language}: {clone_file}")
    except Exception as e:
        print(f"❌ Voice clone failed for {language}: {e}")
    
    # Native pronunciation
    native_file = f"{language}_native.mp3"
    try:
        native_tts = gTTS(text=text, lang=lang_code, slow=False)
        native_tts.save(native_file)
        print(f"✅ Native {language}: {native_file}")
    except Exception as e:
        print(f"❌ Native TTS failed for {language}: {e}")

print("\n🎉 ALL LANGUAGES PROCESSED!")
```

```python
# Download all generated files
import glob

print("📥 Downloading all voice files...")

# Download voice clones
for file in glob.glob("*_your_voice.wav"):
    files.download(file)
    print(f"Downloaded: {file}")

# Download native pronunciations
for file in glob.glob("*_native.mp3"):
    files.download(file)
    print(f"Downloaded: {file}")

print("\n🏆 COMPLETE! You now have your voice in 8+ Indian languages!")
```
