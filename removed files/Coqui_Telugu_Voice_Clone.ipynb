```python
# Upload your voice sample
from google.colab import files
print("📁 Upload your voice sample (input_english.wav):")
uploaded = files.upload()
```

```python
# Voice Cloning Code
import torch
from TTS.api import TTS
import os

# Check GPU
device = "cuda" if torch.cuda.is_available() else "cpu"
print(f"🚀 Using device: {device}")

# Telugu text
telugu_text = "అదే మరి, నీవు లినియర్ రిగ్రెషన్ చూసినప్పుడు, అక్కడ మనం ఒక నంబర్ ప్రిడిక్ట్ చేయడానికి ప్రయత్నిస్తాము. దానిని రిగ్రెషన్ ప్రాబ్లెమ్ అంటారు."

print("📥 Loading XTTS model...")
tts = TTS("tts_models/multilingual/multi-dataset/xtts_v2").to(device)

print("🎯 Cloning your voice...")
# Get uploaded file name
voice_file = list(uploaded.keys())[0]

tts.tts_to_file(
    text=telugu_text,
    speaker_wav=voice_file,
    language="te",
    file_path="telugu_cloned_voice.wav"
)

print("🎉 SUCCESS! Your Telugu voice clone is ready!")
```

```python
# Download your voice clone
from google.colab import files
files.download("telugu_cloned_voice.wav")
print("📥 Download started!")
```

```python
# BONUS: Create multiple Telugu samples
telugu_samples = [
    "నమస్కారం, నేను మీ వాయిస్ క్లోన్",
    "మెషిన్ లర్నింగ్ చాలా ఆసక్తికరమైన విషయం",
    "ఆర్టిఫిషియల్ ఇంటెలిజెన్స్ భవిష్యత్తు"
]

for i, text in enumerate(telugu_samples):
    output_file = f"telugu_sample_{i+1}.wav"
    tts.tts_to_file(
        text=text,
        speaker_wav=voice_file,
        language="te",
        file_path=output_file
    )
    files.download(output_file)
    print(f"✅ Generated: {output_file}")

print("🏆 All samples ready!")
```
